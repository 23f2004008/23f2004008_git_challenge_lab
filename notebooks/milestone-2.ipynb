```python
import subprocess
import os

# Create and manage branches
branches = ['milestone-1', 'milestone-2']

for branch in branches:
    result = subprocess.run(['git', 'branch', branch], cwd=os.getcwd(), capture_output=True, text=True)
    print(f"Branch '{branch}' created or already exists")

# List all branches
result = subprocess.run(['git', 'branch', '-a'], cwd=os.getcwd(), capture_output=True, text=True)
print("\nAvailable branches:")
print(result.stdout)
```
